```python
lst = ['apple',1,'kiwi',2,True]

lst[0:]='watermelon'
print(lst)
```

```python
# Iterating with index

for index , value in enumerate(lst):
    print(f'index is {index} and value is {value}')
print('===end===')
for i in range(len(lst)):
    print(f'index is {i} and value is {lst[i]}')
```
